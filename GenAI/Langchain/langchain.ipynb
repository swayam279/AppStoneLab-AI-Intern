# OpenRouter Model

```python
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter
load_dotenv()
```

```python
model= ChatOpenRouter( model="liquid/lfm-2.5-1.2b-instruct:free", temperature=1)
```

```python
response= model.invoke("What is the acapital of Moon?")
print(response.content)
```

```python
messages = [
    (
        "system",
        "You are a helpful assistant that translates English to French. Translate the user sentence.",
    ),
    ("human", "My name is John."),
]
ai_msg = model.invoke(messages)
ai_msg.content
```

## Streaming

```python
# streaming output

for chunk in model.stream("Write a long poem about the sea."):
    print(chunk.text, end="", flush=True)
```

```python
#Async streaming

async for chunk in model.astream("Write a long poem about the sea."):
    print(chunk.text, end="", flush=True)
```

# Mistral Model

```python
from dotenv import load_dotenv
from langchain_mistralai import ChatMistralAI
load_dotenv()
```

```python
model = ChatMistralAI(
    model="mistral-medium-2508",
    temperature=0.5
)
```

```python
result= model.invoke("What is the capital of Gujarat?")
print(result.content)
```

```python
messages= [
    (
        "system",
        "You are a helpful assistant that always responds by ending the conversation with a new joke with prefix- 'Joke:\n'",
    ),
    ("human", "My name is John."),
]
ai_msg = model.invoke(messages)
print(ai_msg.content)
```

```python
# Image to text

messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": "What's in this image?"
            },
            {
                "type": "image_url",
                "image_url": "https://static.vecteezy.com/system/resources/thumbnails/007/178/392/small/complex-easy-simple-way-from-start-to-idea-chaos-simplifying-problem-solving-and-business-solutions-idea-searching-concept-illustration-vector.jpg"
            }
        ]
    }
]

result= model.invoke(messages)
print(result.content)
```

## Streaming

```python
for output in model.stream("Write a long poem about the sky in a punny tone written by a high school student."):
    print(output.text, end="", flush=True)
```

```python
async for output in model.astream("Write a long poem about the sky in a punny tone written by a high school student."):
    print(output.text, end="", flush=True)
```
