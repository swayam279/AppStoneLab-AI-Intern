```python
from dotenv import load_dotenv
from langchain_mistralai import ChatMistralAI
import streamlit as st
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
load_dotenv()
```

# Simple ChatBot

```python
model= ChatMistralAI(model="mistral-medium-2508")

chat_history=[
    SystemMessage(content='You are a selfless assistant who is always ready to help.')
]

while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        break
    chat_history.append(HumanMessage(content= user_input))
    response = model.invoke(chat_history)
    chat_history.append(AIMessage(content= response.content))
    print("AI:", response.content)

print("Chat History:", chat_history)
```

# ChatBot with Prompt Template

```python
model= ChatMistralAI(model="mistral-medium-2508")

chat_template= ChatPromptTemplate([
    ('system', 'You are the best {domain} expert.'),
    ('human', 'Explain in simple terms, what is {topic}.')    
])
prompt= chat_template.invoke({'domain': 'AI', 'topic': 'Artificial Intelligence'})
print(prompt)
print("\n \n")
response= model.invoke(prompt)
print(response.content)
```

# Message Placeholder

```python
chat_template= ChatPromptTemplate([
    ('system', 'You are a helpful customer support agent.'),
    MessagesPlaceholder(variable_name='chat_history'),
    ('human', '{Query}')
])

chat_history= []
with open('chat_history.txt') as f:
    chat_history.extend(f.readlines())

print(chat_history)
print("\n \n")

prompt= chat_template.invoke({'chat_history': chat_history, 'Query': HumanMessage(content='What is the refund policy?')})
print(prompt)
```
