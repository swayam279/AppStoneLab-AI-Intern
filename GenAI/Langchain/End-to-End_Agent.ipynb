```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_mistralai import ChatMistralAI
from langsmith import Client

load_dotenv()
```

```python
search_tool= DuckDuckGoSearchRun()
results= search_tool.run("Who is the current PM of India?")
print(results)
```

```python
client = Client()
prompt = client.pull_prompt("hwchase17/react")
prompt.template
```

```python
@tool
def search(query: str) -> str:
    """  Searches the web for the answer to a question."""
    return search_tool.run(query)
```

```python
llm= ChatMistralAI(model='mistral-medium-latest')
```

```python
agent= create_agent(
    model= llm,
    tools= [search],
    system_prompt= prompt.template,
)
```

```python
query= "Who is the current PM of India?"
for chunk in agent.stream(
    {"messages": [{"role": "user", "content": query}]},
    stream_mode="updates"  # Streams agent steps: LLM → tools → LLM → final
):
    print(chunk, flush=True)
```

```python
Query= "Who is the current PM of India?"
result= agent.invoke({
    'messages': [
        {'role': 'user', 'content': Query}
    ]
})
result
```
