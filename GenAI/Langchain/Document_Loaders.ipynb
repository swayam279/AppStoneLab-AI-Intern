```python
from pprint import pp

from dotenv import load_dotenv
from langchain_community.document_loaders import CSVLoader, DirectoryLoader, PDFMinerLoader, PyPDFLoader, TextLoader, WebBaseLoader
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_mistralai import ChatMistralAI

load_dotenv()
```

# Text Loaders

```python
loader= TextLoader(r"D:\AppStoneLab\GenAI\Langchain\Documents\Cricket.txt")
docs = loader.load()
print(docs[0])
```

```python
model= ChatMistralAI(model="mistral-medium-2508")

prompt= PromptTemplate(
    template="Summarize the following poem: {poem}",
    input_variables=["poem"]
)

parser= StrOutputParser()

chain= prompt | model | parser

result= chain.invoke({'poem': docs[0].page_content})
print(result)
```
