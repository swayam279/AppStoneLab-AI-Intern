```python
from dotenv import load_dotenv
from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_mistralai import ChatMistralAI, MistralAIEmbeddings

load_dotenv()
```

```python
doc1 = Document(
    page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Playing for Royal Challengers Bangalore, he holds the record for the most runs in a single IPL season and is known for his aggressive batting style, consistency, and leadership.",
    metadata={"team": "Royal Challengers Bangalore"}
)

doc2 = Document(
    page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five IPL titles. Known for his elegant batting and calm leadership, Rohit has played many match-winning innings at the top of the order.",
    metadata={"team": "Mumbai Indians"}
)

doc3 = Document(
    page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing ability, tactical leadership, and exceptional wicketkeeping skills make him one of the greatest players in IPL history.",
    metadata={"team": "Chennai Super Kings"}
)

doc4 = Document(
    page_content="Jasprit Bumrah is widely regarded as one of the best fast bowlers in T20 cricket. Representing Mumbai Indians, he is known for his deadly yorkers, unique bowling action, and ability to deliver under pressure in crucial match situations.",
    metadata={"team": "Mumbai Indians"}
)

doc5 = Document(
    page_content="Andre Russell is one of the most destructive all-rounders in IPL history. Playing for Kolkata Knight Riders, he is famous for his explosive batting in the death overs and his ability to pick crucial wickets with his fast bowling.",
    metadata={"team": "Kolkata Knight Riders"}
)

docs= [doc1, doc2, doc3, doc4, doc5]
```

```python
vector_store= Chroma(
    embedding_function=MistralAIEmbeddings(),
    persist_directory="vectorstore_chroma",
    collection_name="ipl_players",
)

vector_store.add_documents(docs)
```

```python
vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
vector_store.similarity_search(query='Who is a bowler among them?', k=2)
```

```python
vector_store.similarity_search_with_score(query='Who is a bowler among them?', k=2)
```
