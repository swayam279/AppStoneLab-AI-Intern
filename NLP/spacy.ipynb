```python
#standard preprocess pipeline

def is_token_allowed(token):
  return bool(token and str(token).strip and not token.is_stop and not token.is_punct)

def preprocess(token):
  return token.lemma_.strip().lower()

complete_preprocessed_tokens=[preprocess(token) for token in complete_doc if is_token_allowed(token)]
print(complete_preprocessed_tokens)
```

```python
matcher=Matcher(nlp.vocab)

def extract_full_name(nlp_doc):
  pattern = [{"POS": "PROPN"}, {"POS": "PROPN"}]
  matcher.add("FULL_NAME", [pattern])
  matches = matcher(nlp_doc)
  for _, start, end in matches:
    span= nlp_doc[start:end]
    yield span.text

next(extract_full_name(doc))
```
